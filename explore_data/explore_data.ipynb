```python
#수학 점수를 3개 구단으로 나누기 -> 연속형을 범주형으로 변환
#pandas의 cut 함수: 동일 길이로 나눈다.
df_math_cut = pd.cut(df.math, 3)
df_math_cut
```

```python
group1 = df.math.groupby(df_math_cut)
group1.agg(["count", "mean", "std", "min", "max"])
#구간마다 개수, 평균, 표준편차, 최솟값, 최댓값 구하기
```

```python
#pandas의 qcut 함수: 동일 개수로 나눈다.
#labels 인자를 false로 설정하면 (0, 1, 2)구간을 0부터 순차적으로 +1 한다.
df_math_qcut = pd.qcut(df.math, 3, labels=False)
df_math_qcut
```

```python
#구간을 3부터 1까지 -1 한다. <- 이해 못함
#구간을 3부터 1까지 3등분한다.
df_math_qcut = pd.qcut(df.math, 3, labels=np.arange(3, 0, -1))
df_math_qcut
```

```python
group2 = df.math.groupby(df_math_qcut)
grade_1 = group2.get_group(1)
grade_1
```

```python
group2.agg(["count", "mean", "std", "min", "max"])
```
