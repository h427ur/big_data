```python
#데프 간 연결
df1 = pd.DataFrame(
    [
        ['Dog', '3'],
        ['Lamb', '10'],
        ['Tiger', '6'],
        ['Moose', '3']
    ],
    index = ['0', '1', '2', '3'],
    columns = ['동물', '나이']
)

df2 = pd.DataFrame(
    [
        ['Home', '0'],
        ['Park', '2'],
        ['Forest', '0'],
        ['Park', '1'],
    ],
    index = ['0', '1', '2', '3'],
    columns = ['거주지', '뿔의 개수']
)

pd.concat([df1, df2], axis = 1)
#행 기준으로 연결?
```

```python
#데프 간 병합
df3 = pd.DataFrame(
    {
        '고객번호' : [1001, 1002, 1003, 1004, 1005, 1006, 1007],
        '이름' : ['강감찬', '이순신', '홍길동', '장보고', '유관순', '신사임당', '세종대왕']
    },
    columns = ['고객번호', '이름']
)

df4 = pd.DataFrame(
    {
        '고객번호' : [1001, 1001, 1005, 1006, 1008, 1001],
        '금액' : [10000, 20000, 30000, 40000, 50000, 60000]
    },
    columns = ['고객번호', '금액']
)

#inner join
#how 인수는 생략 가능
pd.merge(df3, df4, how = 'inner', on = '고객번호')
```

```python
#full join
#on 인수는 생략 가능
pd.merge(df3, df4, how = 'outer', on = '고객번호')
```

```python
#좌측 병합
pd.merge(df3, df4, how = 'left')
```

```python
#우측 병합
pd.merge(df3, df4, how = 'right')
```

```python
pd.crosstab(tips.sex, tips['day'])
#범주형 변수의 값들을 교차 분석
```

```python
pd.crosstab(tips.sex, tips['day'], margins=True)
#여백 또는 총합(누적 값)
```

```python
pd.crosstab(tips.sex, tips['day']).apply(lambda r: r / len(tips), axis = 1)
#전체 비율 출력
```
