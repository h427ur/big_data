```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_breast_cancer #유방암 데프
```

```python
b_cancer = load_breast_cancer()
b_cancer
```

```python
b_cancer_df = pd.DataFrame(
    b_cancer.data,
    columns=b_cancer.feature_names
)

b_cancer_df
```

```python
b_cancer_df["diagnosis"] = b_cancer.target
b_cancer_df
```

```python
from sklearn.preprocessing import StandardScaler
#정규화
```

```python
scaler = StandardScaler()
b_cancer_scaled = scaler.fit_transform(b_cancer.data)
b_cancer_scaled
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
Y = b_cancer_df.diagnosis
X = b_cancer_scaled

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=.3)
lr_b_cencer = LogisticRegression()
lr_b_cencer.fit(X_train, Y_train)

Y_predict = lr_b_cencer.predict(X_test)
Y_predict
```

```python
from sklearn import metrics
```

```python
score = metrics.accuracy_score(Y_test, Y_predict)
score
#일치율
```
