```python
df = df.dropna(thresh=int(len(df) * .5), axis=1)
msno.matrix(df)
```

```python
df.age = df.age.fillna(df.age.mean())
msno.matrix(df)
```

```python
df.embark_town = df.embark_town.fillna("Southampton")
df.embarked = df.embarked.fillna("S")
msno.bar(df)
```

```python
bins = [1, 20, 60, 100]
df_age = pd.cut(df.age, bins, labels=["미자", "성인", "노인"])
df["age_class"] = df_age
df
```

```python
df["age_class1"] = pd.qcut(df.age, q=3, labels=["미자", "성인", "노인"])
df
```

```python
pd.get_dummies(df)
```

```python
df.groupby(
    ["sex", "age_class", "pclass", "embarked"]
)["survived"].mean()
```
