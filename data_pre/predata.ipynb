```python
from bs4 import BeautifulSoup as bsoup
import requests

#BeautifulSoup 모듈
#   HTTP에서 html을 긁어오는 모듈

#requests 모듈:
#   HTTP에 접근하는 모듈

html = requests.get('https://www.mozilla.org/ko/products/vpn/#pricing')
html
#200이 출력되면 접근이 허용됐다는 의미다.
```

```python
html.text
```

```python
soup = bsoup(html.text, 'html.parser')
tag = soup.findAll('div', {'class' : 'mzp-c-split-body'})
#findAll(내용, 조건) - 리스트로 저장된다?
tag
```

```python
for i in range(0, 3):
    print(tag[i].find('p').text)
```
