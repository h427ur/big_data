```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
```

```python
tips_df = sns.load_dataset("tips")

plt.rc("font", family="NanumGothic")
```

```python
sns.set_palette("pastel")
#deep, muted, bright, pastel, colorblind

sns.stripplot(x="day", y="total_bill", data=tips_df)
plt.title("팁 데이터")
plt.ylabel("지불액")
plt.xlabel("요일")
plt.show()
```

```python
sns.barplot(x="sex", y="total_bill", data=tips_df)
```

```python
sns.barplot(x="sex", y="total_bill", data=tips_df, estimator=len)
```

```python
sns.barplot(x="sex", y="total_bill", data=tips_df, estimator=len, ci="sd")
```

```python
sns.catplot(x="sex", y="total_bill", data=tips_df)
```

```python
sns.boxplot(x="sex", y="total_bill", data=tips_df)
```

```python
sns.boxplot(tips_df)
```

```python
#범주형 단일변수 시각화
sns.countplot(x="day", data=tips_df)
```

```python
sns.catplot(x="day", kind="count", data=tips_df) #이렇게는 안 쓸거 같다
```

```python
#범+연 시각화: catplot, violinplot, swarmplot, stripplot
sns.catplot(x="day", y="total_bill", hue="sex", kind="swarm", data=tips_df)
```

```python
sns.violinplot(x="day", y="total_bill", data=tips_df)
```

```python
sns.swarmplot(x="day", y="total_bill", hue="sex", data=tips_df)
```

```python
#범+범 시각화: heatmap
sexmoker = tips_df.pivot_table(
    index="smoker",
    columns="sex",
    aggfunc="size"
)

sns.heatmap(
    sexmoker,
    cmap=sns.light_palette("black", as_cmap=True),
    annot=True,
    fmt="d",
)
```

```python
#연+연 시각화: jointplot, scatterplot
sns.jointplot(x="total_bill", y="tip", data=tips_df)
```

```python
#연+연 시각화: jointplot, scatterplot
sns.jointplot(x="total_bill", y="tip", data=tips_df, kind="kde")
```

```python
#연+연 시각화: jointplot, scatterplot
sns.scatterplot(x="total_bill", y="tip", data=tips_df)
```

```python
#다변량 시각화
sns.pairplot(tips_df)
```

```python
#다변량 시각화
sns.pairplot(tips_df, hue="sex")
```
