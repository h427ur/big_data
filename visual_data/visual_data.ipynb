```python
import seaborn as sns
```

```python
data = sns.load_dataset("anscombe")
data
```

```python
data_one = data[data.dataset == "I"]
data_two = data[data.dataset == "II"]
data_thr = data[data.dataset == "III"]
data_fou = data[data.dataset == "IV"]
```

```python
import matplotlib.pyplot as plot

plot.plot(data_one.x, data_one.y)
```

```python
fig = plot.figure()

ax1 = fig.add_subplot(2, 2, 1)
ax2 = fig.add_subplot(2, 2, 2)
ax3 = fig.add_subplot(2, 2, 3)
ax4 = fig.add_subplot(2, 2, 4)

ax1.plot(data_one.x, data_one.y, "o")
ax2.plot(data_two.x, data_two.y, "o")
ax3.plot(data_thr.x, data_thr.y, "o")
ax4.plot(data_fou.x, data_fou.y, "o")

ax1.set_title("data_one")
ax2.set_title("data_two")
ax3.set_title("data_thr")
ax4.set_title("data_fou")

fig.tight_layout()
#이게 뭐지?
```

```python
plot.title("linear")
#plot.plot([1, 2, 3, 4, 5], [5, 4, 3, 2, 1], marker="o")
plot.plot([1, 2, 3, 4, 5], label="asc")
plot.plot([5, 4, 3, 2, 1], color="r", label="desc")
plot.plot([2, 2], [5, 1], linestyle="--", label="what")
plot.plot([0, 4], [3, 3], linestyle=":", label="waat")

plot.legend()

plot.xlabel("col")
plot.ylabel("row")

plot.show()
```

```python
plot.title("hawawa")
plot.plot(
    [10, 20, 30, 40],
    [1, 4, 9, 16],
    c="b",
    lw=5, #line width?
    ls="--", #line style
    marker="o",
    ms=15, #marker size
    mec="g", #marker color
    mew="5",
    mfc="red"
)

plot.xlim(0, 50)
plot.ylim(-10, 30)
plot.show()
```
